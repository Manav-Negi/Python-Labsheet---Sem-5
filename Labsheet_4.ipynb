## Q2  (`Q2_linear_regression.py`)

**Implementation:** Trains a one-layer model to learn the relationship y = 2x + 1.

**Expected result: prediction for 5 should be close to 11.**

```python
import tensorflow as tf
import numpy as np

# Training data: y = 2x + 1
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([3, 5, 7, 9], dtype=float)

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer="sgd", loss="mse")
model.fit(x, y, epochs=300, verbose=0)

print("Prediction for x=5:", model.predict(np.array([5.0]), verbose=0)[0][0])
```

## Q4  (`Q4_gradient_tape.py`)

**Implementation:** Uses GradientTape to find the derivative of y = x² at x = 3.

**Output: Gradient: 6.0.**

```python
import tensorflow as tf

x = tf.Variable(3.0)

# Record operations so TensorFlow can calculate the gradient
with tf.GradientTape() as tape:
    y = x ** 2

gradient = tape.gradient(y, x)
print("Gradient:", gradient.numpy())
```

## Q5  (`Q5_custom_loss.py`)

**Implementation:** Defines mean absolute error and passes it to model.compile().

**Expected result: training completes and prints the confirmation.**

```python
import tensorflow as tf

# Mean absolute error as a custom loss function
def custom_loss(y_true, y_pred):
    return tf.reduce_mean(tf.abs(y_true - y_pred))

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer="adam", loss=custom_loss)

x = tf.constant([[1.0], [2.0], [3.0]])
y = tf.constant([[2.0], [4.0], [6.0]])
model.fit(x, y, epochs=5, verbose=0)
print("Training completed with custom loss.")
```

## Q7  (`Q7_sine_neural_network.py`)

**Implementation:** Trains a small neural network using sine values as training data.

**Expected result: prediction at 0 should be close to 0.**

```python
import tensorflow as tf
import numpy as np

# Create sample angles and their sine values
x = np.linspace(-np.pi, np.pi, 500).astype("float32")
y = np.sin(x).astype("float32")

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(32, activation="tanh"),
    tf.keras.layers.Dense(32, activation="tanh"),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.fit(x, y, epochs=100, verbose=0)

print("Predicted sine value at 0:",
      model.predict(np.array([0.0]), verbose=0)[0][0])
```

## Q8  (`Q8_binary_classification.py`)

**Implementation:** Builds a feedforward network with sigmoid output for two-class prediction.

**Expected result: evaluation displays loss and accuracy. Accuracy may vary.**

```python
import tensorflow as tf
import numpy as np

# Simple dataset with two input features and binary labels
x = np.array([[0, 0], [0, 1], [1, 0], [1, 1],
              [0.2, 0.1], [0.8, 0.9]], dtype="float32")
y = np.array([0, 0, 0, 1, 0, 1], dtype="float32")

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(2,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])
model.fit(x, y, epochs=200, verbose=0)

print("Evaluation:", model.evaluate(x, y, verbose=0))
```

## Q10  (`Q10_save_load_model.py`)

**Implementation:** Saves a trained model in Keras format and loads it for prediction.

**Expected result: a numeric prediction close to 8 is printed.**

```python
import tensorflow as tf
import numpy as np

x = np.array([[1.0], [2.0], [3.0]])
y = np.array([[2.0], [4.0], [6.0]])

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer="adam", loss="mse")
model.fit(x, y, epochs=500, verbose=0)

# Save the complete model
model.save("my_model.keras")

# Load it again and make a prediction
loaded_model = tf.keras.models.load_model("my_model.keras")
print("Prediction for x=4 (should be close to 8):",
      loaded_model.predict(np.array([[4.0]]), verbose=0)[0][0])
```

## Q11  (`Q11_callbacks.py`)

**Implementation:** Uses EarlyStopping to stop when loss stops improving and ModelCheckpoint to save the best model.

**Expected result: training finishes and best_model.keras is saved.**

```python
import tensorflow as tf
import numpy as np

x = np.arange(1, 11, dtype="float32")
y = 2 * x

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer="adam", loss="mse")

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="loss", patience=5, restore_best_weights=True
)
checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "best_model.keras", monitor="loss", save_best_only=True
)

model.fit(x, y, epochs=100, callbacks=[early_stop, checkpoint], verbose=0)
print("Training finished. Best model saved.")
```

## Q12  (`Q12_dropout_overfitting.py`)

**Implementation:** Adds dropout, which randomly disables some neurons during training to help reduce overfitting.

**Expected result: training completes. Dropout can reduce overfitting, but does not guarantee better results.**

```python
import tensorflow as tf
import numpy as np

x = np.random.rand(200, 10).astype("float32")
y = (x[:, 0] + x[:, 1] > 1).astype("float32")

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(10,)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])
model.fit(x, y, epochs=10, validation_split=0.2, verbose=0)

print("Training with dropout completed.")
```

## Q13  (`Q13_functional_multiple_io.py`)

**Implementation:** Uses the Functional API to connect two inputs to two output layers.

**Expected result: training completes without errors.**

```python
import tensorflow as tf
import numpy as np

# Two inputs
input_a = tf.keras.Input(shape=(1,), name="first_input")
input_b = tf.keras.Input(shape=(1,), name="second_input")

combined = tf.keras.layers.Concatenate()([input_a, input_b])
hidden = tf.keras.layers.Dense(8, activation="relu")(combined)

# Two outputs
output_sum = tf.keras.layers.Dense(1, name="sum_output")(hidden)
output_product = tf.keras.layers.Dense(1, name="product_output")(hidden)

model = tf.keras.Model(
    inputs=[input_a, input_b],
    outputs=[output_sum, output_product]
)
model.compile(optimizer="adam", loss="mse")

a = np.array([[1.], [2.], [3.], [4.]], dtype="float32")
b = np.array([[2.], [3.], [4.], [5.]], dtype="float32")
sum_y = a + b
product_y = a * b

model.fit([a, b], [sum_y, product_y], epochs=10, verbose=0)
print("Multiple-input, multiple-output training completed.")
```

## Q14  (`Q14_hyperparameter_tuning.py`)

**Implementation:** Uses Keras Tuner RandomSearch to test different neuron counts and learning rates.

**Expected result: the best hyperparameter settings are printed. Install keras-tuner before running.**

```python
# Install first: pip install tensorflow keras-tuner
import tensorflow as tf
import keras_tuner as kt
import numpy as np

x = np.random.rand(200, 4).astype("float32")
y = (x[:, 0] + x[:, 1] > 1).astype("float32")

def build_model(hp):
    model = tf.keras.Sequential()
    model.add(tf.keras.layers.Input(shape=(4,)))
    units = hp.Choice("units", values=[8, 16, 32])
    model.add(tf.keras.layers.Dense(units, activation="relu"))
    model.add(tf.keras.layers.Dense(1, activation="sigmoid"))
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=hp.Choice("learning_rate", values=[0.01, 0.001])
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    return model

tuner = kt.RandomSearch(
    build_model, objective="val_accuracy", max_trials=3,
    directory="tuning_results", project_name="basic_tuning",
    overwrite=True
)
tuner.search(x, y, epochs=5, validation_split=0.2, verbose=0)

print("Best settings:", tuner.get_best_hyperparameters(1)[0].values)
```

## Q15  (`Q15_training_curves.py`)

**Implementation:** Plots training and validation loss and accuracy to compare model performance.

**Expected result: two graphs appear. A widening gap between training and validation curves can indicate overfitting.**

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(300, 4).astype("float32")
y = (x[:, 0] + x[:, 1] > 1).astype("float32")

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(4,)),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])

history = model.fit(
    x, y, epochs=20, validation_split=0.2, verbose=0
)

# Plot loss curves
plt.figure()
plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.title("Training and Validation Loss")
plt.show()

# Plot accuracy curves
plt.figure()
plt.plot(history.history["accuracy"], label="Training accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.title("Training and Validation Accuracy")
plt.show()

print("If training improves while validation worsens, overfitting may be occurring.")
```
