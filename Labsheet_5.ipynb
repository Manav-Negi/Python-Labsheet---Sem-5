## Q2  (`Q2_grayscale_side_by_side.py`)

**Implementation:** Converts a color image to grayscale and places both versions side-by-side.

**Expected result: the original color image and grayscale image appear together.**

```python
import cv2
import numpy as np

image = cv2.imread("input.jpg")
if image is None:
    print("Image not found.")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    # Convert grayscale to 3 channels so both images can be joined
    gray_color = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
    combined = np.hstack((image, gray_color))
    cv2.imshow("Original | Grayscale", combined)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Q3  (`Q3_rgb_hsv_windows.py`)

**Implementation:** Creates RGB and HSV representations and displays them in separate windows. HSV is converted back to BGR for a normal-looking OpenCV display.

**Expected result: two windows appear, one showing the RGB image and the other showing the HSV image converted for viewing.**

```python
import cv2

image = cv2.imread("input.jpg")
if image is None:
    print("Image not found.")
else:
    # Convert OpenCV's BGR image to RGB and HSV
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

    # Convert back to BGR only for correct display in OpenCV windows
    rgb_display = cv2.cvtColor(rgb, cv2.COLOR_RGB2BGR)
    hsv_display = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

    cv2.imshow("RGB Version", rgb_display)
    cv2.imshow("HSV Version (converted for display)", hsv_display)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Q5  (`Q5_rotate_image.py`)

**Implementation:** Rotates the image by 45 degrees around its center.

**Expected result: the image appears rotated. Parts outside the original image boundary may be cropped.**

```python
import cv2

image = cv2.imread("input.jpg")
if image is None:
    print("Image not found.")
else:
    height, width = image.shape[:2]
    center = (width // 2, height // 2)

    # Rotate around the image center by 45 degrees
    matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
    rotated = cv2.warpAffine(image, matrix, (width, height))

    cv2.imshow("Rotated Image", rotated)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Q8  (`Q8_face_detection.py`)

**Implementation:** Uses a pre-trained Haar cascade to detect faces and draws a rectangle around each detected face.

**Expected result: detected faces are marked with green rectangles. Results depend on image quality and face position.**

```python
import cv2

# OpenCV 5.x no longer ships CascadeClassifier; use: pip install "opencv-python<5"
if not hasattr(cv2, "CascadeClassifier"):
    raise SystemExit('This OpenCV version has no CascadeClassifier. Run: python -m pip install "opencv-python<5"')

image = cv2.imread("input.jpg")
if image is None:
    print("Image not found.")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Load the face Haar cascade included with OpenCV
    cascade_path = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
    face_cascade = cv2.CascadeClassifier(cascade_path)

    if face_cascade.empty():
        print("Could not load face cascade.")
    else:
        faces = face_cascade.detectMultiScale(
            gray, scaleFactor=1.1, minNeighbors=5
        )
        for (x, y, w, h) in faces:
            cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)

        print("Faces detected:", len(faces))
        cv2.imshow("Face Detection", image)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
```

## Q9  (`Q9_eye_full_body_detection.py`)

**Implementation:** Detects eyes and full bodies with OpenCV's pre-trained Haar cascade files.

**Expected result: detected eyes and bodies are outlined. Detection may miss objects depending on pose, lighting and image quality.**

```python
import cv2

# OpenCV 5.x no longer ships CascadeClassifier; use: pip install "opencv-python<5"
if not hasattr(cv2, "CascadeClassifier"):
    raise SystemExit('This OpenCV version has no CascadeClassifier. Run: python -m pip install "opencv-python<5"')

image = cv2.imread("input.jpg")
if image is None:
    print("Image not found.")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # OpenCV supplies these pre-trained cascade XML files
    eye_path = cv2.data.haarcascades + "haarcascade_eye.xml"
    body_path = cv2.data.haarcascades + "haarcascade_fullbody.xml"

    eye_cascade = cv2.CascadeClassifier(eye_path)
    body_cascade = cv2.CascadeClassifier(body_path)

    if eye_cascade.empty() or body_cascade.empty():
        print("Could not load one or more cascade files.")
    else:
        eyes = eye_cascade.detectMultiScale(gray, 1.1, 5)
        bodies = body_cascade.detectMultiScale(gray, 1.1, 3)

        # Draw blue rectangles around eyes
        for (x, y, w, h) in eyes:
            cv2.rectangle(image, (x, y), (x + w, y + h), (255, 0, 0), 2)

        # Draw green rectangles around full bodies
        for (x, y, w, h) in bodies:
            cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)

        print("Eyes:", len(eyes), "Full bodies:", len(bodies))
        cv2.imshow("Eye and Full Body Detection", image)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
```

## Q10  (`Q10_geometric_shape_detection.py`)

**Implementation:** Finds contours and estimates each shape using its number of polygon corners and circularity.

**Expected result: triangles, squares, rectangles and circles in a clear high-contrast shapes.jpg image are labelled. Complex or overlapping shapes may be classified incorrectly.**

```python
import cv2

# Use a clear image containing dark shapes on a light background
image = cv2.imread("shapes.jpg")
if image is None:
    print("Image not found.")
else:
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    _, threshold = cv2.threshold(
        blurred, 127, 255, cv2.THRESH_BINARY_INV
    )

    contours, _ = cv2.findContours(
        threshold, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
    )

    for contour in contours:
        area = cv2.contourArea(contour)
        if area < 100:
            continue

        perimeter = cv2.arcLength(contour, True)
        approx = cv2.approxPolyDP(contour, 0.04 * perimeter, True)
        corners = len(approx)

        if corners == 3:
            shape = "Triangle"
        elif corners == 4:
            x, y, w, h = cv2.boundingRect(approx)
            ratio = w / float(h)
            shape = "Square" if 0.9 <= ratio <= 1.1 else "Rectangle"
        else:
            circularity = 4 * 3.14159 * area / (perimeter * perimeter)
            shape = "Circle" if circularity > 0.75 else "Other"

        x, y, w, h = cv2.boundingRect(contour)
        cv2.putText(image, shape, (x, y - 5),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 2)
        cv2.drawContours(image, [contour], -1, (0, 255, 0), 2)

    cv2.imshow("Shape Detection", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Q11  (`Q11_live_webcam.py`)

**Implementation:** Opens the webcam and displays live video until the Q key is pressed.

**Expected result: a live webcam window appears. Press Q to close it.**

```python
import cv2

camera = cv2.VideoCapture(0)
if not camera.isOpened():
    print("Could not open webcam.")
else:
    print("Press Q to stop.")
    while True:
        success, frame = camera.read()
        if not success:
            break

        cv2.imshow("Live Webcam", frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    camera.release()
    cv2.destroyAllWindows()
```

## Q13  (`Q13_save_100_webcam_frames.py`)

**Implementation:** Captures webcam frames and saves up to 100 JPG images in the captured_frames folder.

**Expected result: up to 100 image files are saved. Press Q to stop early; the program reports how many were saved.**

```python
import cv2
import os

camera = cv2.VideoCapture(0)
if not camera.isOpened():
    print("Could not open webcam.")
else:
    os.makedirs("captured_frames", exist_ok=True)
    saved = 0
    print("Capturing 100 frames. Press Q to stop early.")

    while saved < 100:
        success, frame = camera.read()
        if not success:
            print("Could not read a frame.")
            break

        cv2.imshow("Webcam Capture", frame)

        # Save each frame with a unique filename
        filename = f"captured_frames/frame_{saved + 1:03d}.jpg"
        if cv2.imwrite(filename, frame):
            saved += 1

        if cv2.waitKey(30) & 0xFF == ord("q"):
            break

    camera.release()
    cv2.destroyAllWindows()
    print("Frames saved:", saved)
```
